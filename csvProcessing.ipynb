```python
import pandas as pd

loadedCsv = pd.read_excel('new.xlsx')
```

```python
from queryMatching import normalize

cleanedCsv = pd.DataFrame()

cleanedCsv['game'] = loadedCsv['OA.']
cleanedCsv['songName'] = loadedCsv['Song name']
cleanedCsv['ContributingArtists'] = loadedCsv['Artist']
cleanedCsv['difficulty'] = loadedCsv['Difficulty']
cleanedCsv['coaches'] = loadedCsv['Mode']
cleanedCsv['alternate'] = loadedCsv['Alternate']
cleanedCsv["JD+"] = ~loadedCsv["JD+"].eq("X")

cleanedCsv['extreme'] = cleanedCsv['difficulty'].str.contains('Extreme', case=False, na=False)
cleanedCsv['extreme'] = cleanedCsv['alternate'].str.contains('Extreme', case=False, na=False) | cleanedCsv['extreme']

print('saving to csv...')

cleanedCsv.to_csv('resources/songLists/JustDanceSongsCleaned.csv', index=False)

df = cleanedCsv.copy()

df['songs_normalized'] = df['songName'].fillna("").apply(normalize)
df['artists_normalized'] = df['ContributingArtists'].fillna("").apply(normalize)

df["combined_norm"] = (df["songs_normalized"] + " " + df["artists_normalized"])
df["combined_norm_inv"] = (df["artists_normalized"] + " " + df["songs_normalized"])
df.to_csv('songsNew.csv', index=False)
```

```python
from cryptography.fernet import Fernet

key = Fernet.generate_key()
print(key)
cipher = Fernet(key)

with open("resources/songLists/songs_fixed.csv", "rb") as f:
    plaintext = f.read()

encrypted = cipher.encrypt(plaintext)

with open("resources/songLists/data.csv.enc", "wb") as f:
    f.write(encrypted)
```

```python
import json
import pandas as pd

df = pd.read_csv('resources/songLists/songs.csv')

alternates_df = df[df['alternate'].notna() & (df['alternate'] != '')]

# Remove unwanted rows
alternates_df = alternates_df[
    ~alternates_df['alternate'].str.contains('extreme', case=False, na=False)
]
alternates_df = alternates_df[
    ~alternates_df['alternate'].str.contains('-', case=False, na=False)
]

# Normalize for duplicate removal (important!)
alternates_df['alt_norm'] = alternates_df['alternate'].str.lower().str.strip()

# Remove duplicates based on normalized form
alternates_df = alternates_df.drop_duplicates(subset=['alt_norm'])

# Sort longest to shortest
alternates_df = alternates_df.sort_values(
    by='alternate',
    key=lambda col: col.str.len(),
    ascending=False
)

alternates = alternates_df['alternate'].tolist()

with open('resources/songLists/alternates.json', 'w', encoding='utf-8') as f:
    json.dump(alternates, f, indent=2, ensure_ascii=False)

print(f'Saved {len(alternates)} alternates to resources/songLists/alternates.json')
```

```python
import pandas as pd

input_file = "resources/songLists/songs.csv"
output_file = "resources/songLists/songs_fixed.csv"

# Read raw text, replace curly quote with straight apostrophe, write back
with open(input_file, "r", encoding="utf-8") as f:
    content = f.read()

content = content.replace("’", "'")

with open(output_file, "w", encoding="utf-8") as f:
    f.write(content)

print(f"Done. Saved to {output_file}")
```
